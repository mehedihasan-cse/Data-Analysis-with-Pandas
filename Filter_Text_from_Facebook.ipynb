```python
import pandas as pd
```

```python
df = pd.read_csv('/content/ESUIT _ Comments Exporter for Facebook™ (125).csv')
```

```python
df.head(2)
```

```python
df.columns
```

```python
df = df.drop(columns = ['Id', 'UserId', 'Avatar', 'Author', 'Url', 'ReactionsCount',
       'Depth', 'SubCommentsCount', 'CommentAt'])
df
```

```python
df
```

```python
df.isna().sum()
```

```python
import re

def has_english(row):
    # Check if the 'Content' column contains any English letters
    return bool(re.search(r'[a-zA-Z]', row['Content']))

# Filter the DataFrame to keep only rows without English letters
df = df[~df.apply(has_english, axis=1)]

print(df)
```

```python
df.isna().sum()
```

```python
df.reset_index(drop=True, inplace=True)
```
